```
print("Name :- Vanshika")
print("Roll no :- 50")
print("Course :- BCA")
print("Section :- H1")
print("-"*40)

import seaborn as sns
df = sns.load_dataset('titanic')

fare_range = df['fare'].max() - df['fare'].min()
Q1 = df['fare'].quantile(0.25)
Q3 = df['fare'].quantile(0.75)
IQR = Q3 - Q1
MAD = (df['fare'] - df['fare'].mean()).abs().mean()

print("Range:", fare_range)
print("IQR:", IQR)
print("MAD:", MAD)
```

```
print("Name :- Vanshika")
print("Roll no :- 50")
print("Course :- BCA")
print("Section :- H1")
print("-"*40)

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('titanic')

plt.boxplot(df['fare'].dropna())
plt.title("Boxplot of Fare")
plt.show()
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['fare'] < lower) | (df['fare'] > upper)]

print("Number of Outliers:", outliers.shape[0])

df_no_outliers = df[(df['fare'] >= lower) & (df['fare'] <= upper)]

print("Mean Fare Before:", df['fare'].mean())
print("Mean Fare After:", df_no_outliers['fare'].mean())

```

```
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.histplot(df['fare'], kde=True)
plt.title("Before Removing Outliers")

plt.subplot(1,2,2)
sns.histplot(df_no_outliers['fare'], kde=True)
plt.title("After Removing Outliers")
plt.show()

```

```
plt.scatter(df['age'], df['fare'])
plt.axhline(df['fare'].mean(), color='r', linestyle='--')
plt.axvline(df['age'].mean(), color='g', linestyle='--')

plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.show()
df_mean = df.copy()
df_median = df.copy()

df_mean['age'].fillna(df['age'].mean(), inplace=True)
df_median['age'].fillna(df['age'].median(), inplace=True)

print("Mean after Mean Imputation:", df_mean['age'].mean())
print("Mean after Median Imputation:", df_median['age'].mean())
```
